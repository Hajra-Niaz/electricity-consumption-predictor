```python
import pandas as pd
```

```python
df=pd.read_csv('electricity_consumption_forecast.csv')
```

```python
df
```

```python
x=df[['Month','Avg_Temperature_C','Residents','Past_Month_Consumption_kWh']]
y=df['Next_Month_Consumption_kWh']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

Apply Model:

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
model.predict(x_test)
```

Apply Joblib to Load Model:

```python
import joblib as jb
jb.dump(model,'elec.pkl')
```
